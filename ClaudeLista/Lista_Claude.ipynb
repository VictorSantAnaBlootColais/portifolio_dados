#### 1. Contador de palavras: leia um texto e mostre as 5 palavras mais frequentes (use dict ou collections.Counter)

```python
Frase = "Se a classe trabalhadora produz tudo a ela tudo pertence"
palavras = Frase.split()
contador = {}
mais_frequentes = {}
top5_maisfrequentes = 0

for palavra in palavras:
    if palavra in contador:
        contador[palavra] += 1
    else:
        contador[palavra] = 1

mais_frequentes = dict(sorted(contador.items(), key=lambda item: item[1], reverse= True))
for palavra, vezes in mais_frequentes.items():
    if top5_maisfrequentes < 5:
        print(f"{palavra} : {vezes}")
        top5_maisfrequentes += 1

```
